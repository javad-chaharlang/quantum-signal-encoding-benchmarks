# Basis-Encoded Quantum Audio Representation

This notebook prepares, verifies, simulates, and reconstructs a small quantized audio signal. The method is an educational basis-encoding baseline and is **not** labeled QRDA.

```python
from qseb.audio import (
    build_basis_encoded_audio_circuit,
    circuit_resource_metrics,
    exact_basis_probabilities,
    reconstruct_from_counts,
    simulate_counts,
)
```

```python
samples = [3, 6, 2, 5]
circuit, spec = build_basis_encoded_audio_circuit(samples, amplitude_bits=3)
print(spec)
circuit.draw('mpl')
```

```python
exact_basis_probabilities(circuit, spec)
```

```python
counts = simulate_counts(circuit, shots=4096, seed_simulator=42)
reconstructed = reconstruct_from_counts(counts, spec)
print('Original:', samples)
print('Reconstructed:', reconstructed)
```

```python
circuit_resource_metrics(circuit)
```

## Interpretation

Exact reconstruction confirms circuit correctness for this deterministic basis representation. Future experiments will study shot coverage, transpiled depth, and noise sensitivity as the number of samples and amplitude bits increases.
